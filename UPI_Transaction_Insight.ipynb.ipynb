```python
import numpy as np
```

```python
conditions = [
    transactions["amount_deviation_score"] < 0.5,
    transactions["amount_deviation_score"] < 1,
    transactions["amount_deviation_score"] < 2,
    transactions["amount_deviation_score"] < 3
]

labels = [
    "Low",
    "Moderate",
    "High",
    "Very High"
]

transactions["deviation_group"] = np.select(
    conditions,
    labels,
    default="Extreme"
)

fraud_by_deviation = (
    transactions.groupby("deviation_group", observed=True)["is_fraud"]
    .mean() * 100
)

fraud_by_deviation
```

Fraud rates generally increase as transaction amount deviation rises, reaching 5.83% for very-high deviation scores (2–3). However, the extreme deviation group shows a lower rate of 4.90%, indicating that the relationship is not strictly linear.

```python
fraud_by_combined_risk = (
    transactions.groupby(
        ["new_device_flag", "ip_location_mismatch"]
    )["is_fraud"]
    .mean() * 100
)

fraud_by_combined_risk
```

Transactions involving both a new device and an IP location mismatch showed the highest observed fraud rate (16.67%), compared with 2.98% for transactions with neither signal. This suggests that the combination of these two risk indicators may be particularly useful for identifying potentially fraudulent transactions

```python
fraud_by_failed_attempts = (
    transactions.groupby("failed_attempts_last_24h")["is_fraud"]
    .mean() * 100
)

fraud_by_failed_attempts.sort_values(ascending=False)
```

Fraud rates generally increased with the number of failed attempts in the previous 24 hours. Transactions associated with four failed attempts had the highest observed fraud rate at 8.42%, compared with 3.46% for transactions with no failed attempts.

```python
transactions["high_failed_attempts"] = (
    transactions["failed_attempts_last_24h"] >= 3
)

fraud_by_risk = (
    transactions.groupby(
        ["new_device_flag", "ip_location_mismatch", "high_failed_attempts"]
    )["is_fraud"]
    .mean() * 100
)

fraud_by_risk.sort_values(ascending=False)
```

The combination of multiple fraud-risk indicators produced substantially higher observed fraud rates. Transactions involving a new device, IP location mismatch, and three or more failed attempts had a 50% observed fraud rate, compared with 2.79% when none of these indicators were present. This highlights the potential value of combining multiple behavioral and security signals for fraud-risk identification.


```python
risk_summary = (
    transactions.groupby(
        ["new_device_flag", "ip_location_mismatch", "high_failed_attempts"]
    )["is_fraud"]
    .agg(
        transaction_count="count",
        fraud_count="sum",
        fraud_rate="mean"
    )
)

risk_summary["fraud_rate"] = risk_summary["fraud_rate"] * 100

risk_summary.sort_values("fraud_rate", ascending=False)
```

Combining multiple fraud-risk indicators reveals substantially higher fraud rates than transactions without these indicators. While the highest observed rate was 50%, it was based on only two transactions. More reliable groups also showed elevated fraud rates, particularly transactions involving a new device, where fraud rates exceeded 11%. This suggests that combining risk signals can provide stronger fraud-risk identification than relying on individual indicators alone.
